###This notebook ingests raw COVID-19 and vaccination datasets into
###Databricks as the Bronze layer.

In this step, original CSV datasets are loaded into Databricks without any modification and stored in a governed storage location.  
The Bronze layer preserves raw data integrity and serves as the foundation for downstream cleaning, transformation, and AI workflows.


### Read the COVID CSV file

```python
covid_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/workspace/default/raw_data/COVID-19_India_Data.csv")
)
```

###Read the Vaccination CSV file


```python
vaccine_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/workspace/default/raw_data/Vaccination.csv")
)
```
